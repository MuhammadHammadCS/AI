```python
stdict = {"name":"Emma","marks":85,"grade": ''}
print(stdict)
if(stdict["marks"] >= 80):
    stdict["grade"] = 'A'
elif(stdict["marks"] >= 70):
    stdict["grade"] = 'B'
elif(stdict["marks"] >= 60):
    stdict["grade"] = 'C'
else:
    stdict["grade"] = 'U'

print(stdict)
```

```python
import math

num = int(input("Enter a number: "))
out = "true"
if(num < 2):
    out = "false"
for i in range(2,int(math.sqrt(num) + 1)):
    if(num % i == 0):
        out = "false"
print("Number: ",num)
print("Is prime: ",out)
```

```python
def find_max(*args):
    return max(args)

def print_key_values(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")

print(find_max(10, 25, 5, 40, 15))

print_key_values(name="Ali", age=26, city="Alaska")
```
