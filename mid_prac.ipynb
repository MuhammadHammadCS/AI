```python
#DLS
graph = {
'A': ['B', 'C'],
'B': ['D', 'E'],
'C': ['F', 'G'],
'D': ['H'],
'E': [],
'F': ['I'],
'G': [],
'H': [],
'I': []
}

def dls(graph, start, goal, depth_limit):
    def dfs(node, goal, depth, path):
        if depth > depth_limit:
            return None
        path.append(node)
        if node == goal:
            print(f"Path found using DLS path: {path}")
            return
        for neighbor in graph.get(node, []):
            if neighbor not in path:
                result = dfs(neighbor,goal, depth+1,path)
                if result:
                    return result
        path.pop()
        return None
    return dfs(start,goal,0,[])


dls(graph,'A','D',3)
```

```python
graph = {
'A': {'B': 2, 'C': 1},
'B': {'D': 4, 'E': 3},
'C': {'F': 1, 'G': 5},
'D': {'H': 2},
'E': {},
'F': {'I': 6},
'G': {},
'H': {},
'I': {}
}

def ucs(graph, start, goal):
    frontier = [ (start, 0) ]
    visited = set()
    cost_so_far = { start: 0 }
    came_from = { start: None }

    while frontier:
        frontier.sort(key=lambda x: x[1])

        curr_node, curr_cost = frontier.pop(0)

        if curr_node in visited:
            continue

        visited.add(curr_node)

        if curr_node == goal:
            path = []
            while curr_node is not None:
                path.append(curr_node)
                curr_node = came_from[curr_node]
            path.reverse()
            print(f"Path found in UCS . Path: {path}, Total cost: {curr_cost}")
            return

        for neighbor, cost in graph[curr_node].items():
            new_cost = curr_cost + cost
            if neighbor not in cost_so_far or new_cost < cost_so_far[neighbor]:
                cost_so_far[neighbor] = new_cost
                came_from[neighbor] = curr_node
                frontier.append((neighbor, new_cost))
    print("Goal Not Found")

ucs(graph, 'A', 'I')
```
